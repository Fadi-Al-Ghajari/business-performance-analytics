Install the libraries

```python
import pandas as pd
import numpy as np
import random

from faker import Faker

fake = Faker()
np.random.seed(42)
random.seed(42)
```

Generate regions

```python
regions = pd.DataFrame({
    "Region_ID": ["R01", "R02", "R03", "R04", "R05"],
    "Region": [
        "North",
        "South",
        "East",
        "West",
        "Central"
    ]
})

regions
```

Generate customers

```python
n_customers = 5000

segments = [
    "Consumer",
    "Corporate",
    "Small Business"
]

customers = pd.DataFrame({
    "Customer_ID": [
        f"CUST{i:05d}"
        for i in range(1, n_customers + 1)
    ],

    "Customer_Name": [
        fake.name()
        for _ in range(n_customers)
    ],

    "Segment": np.random.choice(
        segments,
        size=n_customers,
        p=[0.60, 0.20, 0.20]
    ),

    "Region_ID": np.random.choice(
        regions["Region_ID"],
        size=n_customers,
        p=[0.20, 0.18, 0.22, 0.25, 0.15]
    ),

    "Customer_Since": pd.to_datetime(
        np.random.choice(
            pd.date_range(
                "2019-01-01",
                "2024-12-31"
            ),
            size=n_customers
        )
    )
})

customers.head()
```

Generate products

```python
categories = {
    "Technology": [
        "Laptops",
        "Monitors",
        "Accessories",
        "Phones"
    ],

    "Furniture": [
        "Chairs",
        "Desks",
        "Storage",
        "Tables"
    ],

    "Office Supplies": [
        "Paper",
        "Writing",
        "Binders",
        "Stationery"
    ],

    "Home": [
        "Kitchen",
        "Decor",
        "Lighting",
        "Appliances"
    ]
}
```

```python
n_products = 200

product_rows = []

for i in range(1, n_products + 1):

    category = random.choice(
        list(categories.keys())
    )

    sub_category = random.choice(
        categories[category]
    )

    unit_cost = round(
        np.random.uniform(5, 800),
        2
    )

    markup = np.random.uniform(
        1.15,
        2.2
    )

    list_price = round(
        unit_cost * markup,
        2
    )

    product_rows.append({
        "Product_ID": f"PROD{i:04d}",
        "Product_Name": f"{sub_category} Product {i:03d}",
        "Category": category,
        "Sub_Category": sub_category,
        "Unit_Cost": unit_cost,
        "List_Price": list_price
    })

products = pd.DataFrame(product_rows)

products.head()
```

Generate orders

```python
n_orders = 30000

orders = pd.DataFrame({
    "Order_ID": [
        f"ORD{i:06d}"
        for i in range(1, n_orders + 1)
    ],

    "Order_Date": pd.to_datetime(
        np.random.choice(
            pd.date_range(
                "2022-01-01",
                "2025-12-31"
            ),
            size=n_orders
        )
    ),

    "Customer_ID": np.random.choice(
        customers["Customer_ID"],
        size=n_orders
    ),

    "Sales_Channel": np.random.choice(
        ["Online", "Retail"],
        size=n_orders,
        p=[0.65, 0.35]
    ),

    "Payment_Method": np.random.choice(
        [
            "Credit Card",
            "Debit Card",
            "Bank Transfer",
            "PayPal"
        ],
        size=n_orders,
        p=[0.45, 0.25, 0.15, 0.15]
    )
})
```

```python
orders = orders.merge(
    customers[
        ["Customer_ID", "Region_ID"]
    ],
    on="Customer_ID",
    how="left"
)

orders.head()
```

Generate order items

```python
n_items = 60000

order_items = pd.DataFrame({
    "Order_Item_ID": [
        f"ITEM{i:07d}"
        for i in range(1, n_items + 1)
    ],

    "Order_ID": np.random.choice(
        orders["Order_ID"],
        size=n_items
    ),

    "Product_ID": np.random.choice(
        products["Product_ID"],
        size=n_items
    ),

    "Quantity": np.random.choice(
        [1, 2, 3, 4, 5, 6, 7, 8],
        size=n_items,
        p=[
            0.35,
            0.25,
            0.15,
            0.10,
            0.06,
            0.04,
            0.03,
            0.02
        ]
    ),

    "Discount": np.random.choice(
        [0, 0.05, 0.10, 0.15, 0.20, 0.30],
        size=n_items,
        p=[
            0.25,
            0.20,
            0.25,
            0.15,
            0.10,
            0.05
        ]
    )
})

order_items.head()
```

Create the analytical sales table

```python
sales = (
    order_items
    .merge(
        orders,
        on="Order_ID",
        how="left"
    )
    .merge(
        products,
        on="Product_ID",
        how="left"
    )
    .merge(
        regions,
        on="Region_ID",
        how="left"
    )
)
```

```python
sales["Revenue"] = (
    sales["Quantity"]
    * sales["List_Price"]
    * (1 - sales["Discount"])
)

sales["Cost"] = (
    sales["Quantity"]
    * sales["Unit_Cost"]
)

sales["Profit"] = (
    sales["Revenue"]
    - sales["Cost"]
)

sales["Profit_Margin"] = (
    sales["Profit"]
    / sales["Revenue"]
)
```

```python
sales[
    [
        "Order_ID",
        "Product_Name",
        "Quantity",
        "Discount",
        "Revenue",
        "Cost",
        "Profit",
        "Profit_Margin"
    ]
].head()
```

Now deliberately introduce data-quality problems

```python
duplicate_rows = sales.sample(
    100,
    random_state=42
)

sales_raw = pd.concat(
    [sales, duplicate_rows],
    ignore_index=True
)
```

```python
missing_customer_idx = sales_raw.sample(
    75,
    random_state=10
).index

sales_raw.loc[
    missing_customer_idx,
    "Customer_ID"
] = np.nan
```

```python
missing_discount_idx = sales_raw.sample(
    50,
    random_state=20
).index

sales_raw.loc[
    missing_discount_idx,
    "Discount"
] = np.nan
```

```python
technology_idx = sales_raw[
    "Category"
] == "Technology"

sample_idx = sales_raw[
    technology_idx
].sample(
    40,
    random_state=30
).index

sales_raw.loc[
    sample_idx,
    "Category"
] = "technology"
```

Save the raw dataset
