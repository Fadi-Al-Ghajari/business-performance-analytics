Recalculate the financial metrics

```python
df["Revenue"] = (
    df["Quantity"]
    * df["List_Price"]
    * (1 - df["Discount"])
)

df["Cost"] = (
    df["Quantity"]
    * df["Unit_Cost"]
)

df["Profit"] = (
    df["Revenue"]
    - df["Cost"]
)

df["Profit_Margin"] = (
    df["Profit"]
    / df["Revenue"]
)
```

Now deal with the 100 duplicates

```python
df = df.drop_duplicates().copy()
```

Validate financial calculations

```python
revenue_check = (
    df["Quantity"]
    * df["List_Price"]
    * (1 - df["Discount"])
)

profit_check = revenue_check - (
    df["Quantity"] * df["Unit_Cost"]
)

print(
    "Revenue calculation errors:",
    (~np.isclose(df["Revenue"], revenue_check)).sum()
)

print(
    "Profit calculation errors:",
    (~np.isclose(df["Profit"], profit_check)).sum()
)
```
